**Resumen: Video 2**

Las **CNN** se usan para que una computadora pueda analizar imágenes. Trabajan con un **campo receptivo local**, donde las neuronas observan pequeñas partes de la imagen. Se define un **filtro**, se aplica una **zancada (stride)** y se puede añadir **relleno (padding)** para controlar la dimensionalidad. Luego, el **pooling**, especialmente el **max pooling**, reduce los datos conservando los detalles más importantes.

Con la evolución de los modelos **CNN**, se han logrado tareas más avanzadas como la **segmentación de imágenes**, donde se identifica cada parte de una imagen. En general, estas redes tienen muchas aplicaciones y su uso depende de lo que se quiera lograr: **el límite es la imaginación**.
