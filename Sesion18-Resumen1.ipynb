**Resumen: Video 1**

TensorFlow permite trabajar con **tensores**, que son los datos básicos de la IA. Los tensores estándar son **inmutables**, mientras que `tf.Variable` permite modificar valores durante el entrenamiento. Para aprender, TensorFlow usa la **autodiferenciación en modo inverso**, calculando **gradientes** mediante la **regla de la cadena** y recorriendo las operaciones hacia atrás.

También utiliza **grafos de computación**, que representan las operaciones como una estructura organizada y abstracta. Esto permite controlar y optimizar el proceso, mejorar la **estabilización** y crear sistemas más robustos. Además, se pueden hacer pruebas o muestras y tener mayor **control manual** sobre los cálculos. Todo esto se puede trabajar directamente con TensorFlow, sin depender de bibliotecas externas.
